```python
class Employee:
    def __init__(self, employeeId, name, salary):
        self.__employeeId = employeeId
        self.__name = name
        self.__salary = salary

    def get_employeeId(self):
        return self.__employeeId

    def set_employeeId(self, employeeId):
        self.__employeeId = employeeId

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        self.__salary = salary

emp1 = Employee("EMP101", "harshvardhan", 75000.00)
print(f"{emp1.get_employeeId()},{emp1.get_name()},{emp1.get_salary()}")

emp1.set_name("harshvardhan")
emp1.set_salary("75000.00")

print (f"update the name and salary : \n{emp1.get_name()},\n{emp1.get_salary()}")
```
